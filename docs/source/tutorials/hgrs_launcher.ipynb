```python

import os
import glob

import pandas as pd

from datetime import datetime, timedelta

opj = os.path.join
```

### For PRISMA

```python

idir='/data/satellite/prisma/touria'

odir='/data/satellite/prisma/touria'
cams_dir = '/data/cams/world/'
sites=['bassas','glorieuse','mayotte']

idir='/data/satellite/prisma/L1'
odir='/data/satellite/prisma/L2A'
sites=['actium']
resolution=60
cc_max = 0.6
no_clobber = True


# --------------------
# set options
options=''
if no_clobber:
    options = options+' --no_clobber'


# ------------------
# setting up loop on files available

for site in sites: 
   
    kwarg = ''
    
    # ------------------
    # check if L1C exists and set directories / files
   
    
    l1c_dir = opj(idir,site)
    if not os.path.exists(l1c_dir):
        continue
        
    l2a_dir = opj(odir,site)
    

    files =glob.glob(opj(l1c_dir,'PRS_L1_STD_OFFL_*.he5'))
    
    for file in  files:
        basename= os.path.basename(file)
        date = basename.split('_')[4]
        year, month = date[0:4],date[4:6]
        date_str=year+'-'+month
        cams_file = opj(cams_dir,'cams_forecast_'+date_str+'.nc')
        print(f'hgrs_prisma {file}  --cams_file {cams_file} --odir {l2a_dir} '+options)
```

```python

idir='/data/satellite/prisma/touria'

odir='/data/satellite/prisma/touria'
cams_dir = '/data/cams/world/'
sites=['bassas','glorieuse','mayotte']

idir='/data/satellite/enmap/EnMAP_L1C'
odir='/data/satellite/enmap/L2A'
sites=['AAOT']
sites=os.listdir(idir)
sites = ['senegal']
resolution=60
cc_max = 0.6
no_clobber = True


# --------------------
# set options
options=''
if no_clobber:
    options = options+' --no_clobber'


# ------------------
# setting up loop on files available

for site in sites: 
   
    kwarg = ''
    
    # ------------------
    # check if L1C exists and set directories / files
   
    
    l1c_dir = opj(idir,site)
    if not os.path.exists(l1c_dir):
        continue
        
    l2a_dir = opj(odir,site)
    

    files =glob.glob(opj(l1c_dir,'ENMAP*Z'))
    files.sort()
    for file in  files:
        basename= os.path.basename(file)
        date = basename.split('_')[5]
        year, month = date[0:4],date[4:6]
        date_str=year+'-'+month
        cams_file = opj(cams_dir,'cams_forecast_'+date_str+'.nc')
        print(f'hgrs_enmap {file}  --cams_file {cams_file} --odir {l2a_dir} '+options)
```
